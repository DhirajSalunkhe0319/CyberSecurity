```python
import random
import string
import pytesseract
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt
```

```python
captcha = ''.join(random.choices(string.ascii_uppercase + string.digits+string.ascii_lowercase, k=5))

img = Image.new("RGB", (200, 80), "white")
draw = ImageDraw.Draw(img)
draw.text((50, 25), captcha, fill="black")

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
text = pytesseract.image_to_string(img)

print("Original CAPTCHA:", captcha)
print("OCR Prediction:", text.strip())
```

```python
if captcha == text.strip():
    print("CAPTCHA Broken Successfully!")
else:
    print("CAPTCHA Not Broken!")
```

```python
user_input = input("Enter the CAPTCHA: ")

if user_input == captcha:
    print("CAPTCHA Verified Successfully!")
else:
    print("Incorrect CAPTCHA!")
```
